```python
transactions = []

# Keywords used to auto-detect a category from the description.
# Plain dictionary -- no classes, no fancy matching.
CATEGORY_KEYWORDS = {
    'food': 'Food', 'grocery': 'Food', 'restaurant': 'Food', 'zomato': 'Food', 'swiggy': 'Food',
    'rent': 'Rent', 'landlord': 'Rent',
    'netflix': 'Subscription', 'spotify': 'Subscription', 'prime': 'Subscription',
    'uber': 'Transport', 'ola': 'Transport', 'fuel': 'Transport', 'petrol': 'Transport',
    'electricity': 'Utilities', 'water bill': 'Utilities', 'wifi': 'Utilities',
    'salary': 'Income', 'credited': 'Income'
}


def categorize(description):
  description = description.lower()
  for keyword in CATEGORY_KEYWORDS:
    if keyword in description:
      return CATEGORY_KEYWORDS[keyword]
  return 'Other'


def add_transaction():
  date = input('Enter the Date (YYYY-MM-DD): ')
  description = input('Enter the Description: ')
  amount = float(input('Enter the Amount: '))

  category = categorize(description)

  transaction = {
      'Date': date,
      'Description': description,
      'Amount': amount,
      'Category': category
  }

  transactions.append(transaction)
  print(f"Transaction Added Successfully -> Category detected: {category}")


def view_transactions():
  if not transactions:
    print("No Transaction Records Found!....")
    return

  print('\n ----- Transaction Records --------')

  for i in transactions:
    print("Date →", i['Date'])
    print("Description →", i['Description'])
    print("Amount →", i['Amount'])
    print("Category →", i['Category'])
    print("-" * 30)


def category_summary():
  if not transactions:
    print("No Transaction Records Found!....")
    return

  totals = {}
  for i in transactions:
    cat = i['Category']
    amt = i['Amount']
    if cat in totals:
      totals[cat] = totals[cat] + amt
    else:
      totals[cat] = amt

  print('\n ----- Spending by Category --------')
  for cat in totals:
    print(f"{cat} → {totals[cat]:.2f}")


def monthly_summary():
  if not transactions:
    print("No Transaction Records Found!....")
    return

  totals = {}
  for i in transactions:
    month = i['Date'][:7]  # YYYY-MM part of the date string
    amt = i['Amount']
    if month in totals:
      totals[month] = totals[month] + amt
    else:
      totals[month] = amt

  print('\n ----- Spending by Month --------')
  for month in totals:
    print(f"{month} → {totals[month]:.2f}")


def detect_recurring():
  if not transactions:
    print("No Transaction Records Found!....")
    return

  seen = {}
  for i in transactions:
    key = i['Description'].lower().strip()
    if key in seen:
      seen[key] = seen[key] + 1
    else:
      seen[key] = 1

  print('\n ----- Possible Recurring Charges --------')
  found_any = False
  for description in seen:
    if seen[description] >= 2:
      print(f"{description} → appeared {seen[description]} times")
      found_any = True

  if not found_any:
    print("No recurring charges detected yet.")


def show_insights():
  if not transactions:
    print("No Transaction Records Found!....")
    return

  total_spent = 0
  for i in transactions:
    total_spent = total_spent + i['Amount']

  average_spent = total_spent / len(transactions)

  print('\n ----- Insights Summary --------')
  print(f"Total Transactions   : {len(transactions)}")
  print(f"Total Amount Spent   : {total_spent:.2f}")
  print(f"Average Transaction  : {average_spent:.2f}")


def save_to_file():
  filename = input('Enter filename to save (e.g. records.txt): ')
  file = open(filename, 'w')

  for i in transactions:
    line = i['Date'] + ',' + i['Description'] + ',' + str(i['Amount']) + ',' + i['Category']
    file.write(line + '\n')

  file.close()
  print("Records Saved Successfully")


def load_from_file():
  filename = input('Enter filename to load (e.g. records.txt): ')

  try:
    file = open(filename, 'r')
  except FileNotFoundError:
    print("File not found!")
    return

  for line in file:
    line = line.strip()
    parts = line.split(',')
    if len(parts) == 4:
      transaction = {
          'Date': parts[0],
          'Description': parts[1],
          'Amount': float(parts[2]),
          'Category': parts[3]
      }
      transactions.append(transaction)

  file.close()
  print("Records Loaded Successfully")


def main():
  while True:
    print('\n===== Personal Finance Analyzer =====')
    print('1. Add Transaction')
    print('2. View All Transactions')
    print('3. Category-wise Summary')
    print('4. Monthly Summary')
    print('5. Detect Recurring Charges')
    print('6. Show Insights')
    print('7. Save Records to File')
    print('8. Load Records from File')
    print('9. Exit')

    choice = input('Enter your choice: ')

    if choice == '1':
      add_transaction()
    elif choice == '2':
      view_transactions()
    elif choice == '3':
      category_summary()
    elif choice == '4':
      monthly_summary()
    elif choice == '5':
      detect_recurring()
    elif choice == '6':
      show_insights()
    elif choice == '7':
      save_to_file()
    elif choice == '8':
      load_from_file()
    elif choice == '9':
      print("Exiting... Thank you!")
      break
    else:
      print("Invalid choice, please try again.")


if __name__ == "__main__":
  main()
```
